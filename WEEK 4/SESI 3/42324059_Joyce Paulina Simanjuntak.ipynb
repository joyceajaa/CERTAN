# Naïve Bayes  

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Cell ini digunakan untuk mengimpor library yang dibutuhkan, yaitu NumPy, Pandas, dan load_iris dari Scikit-learn. NumPy dan Pandas digunakan untuk membantu pengolahan data, sedangkan load_iris berfungsi untuk memuat Dataset Iris yang berisi karakteristik tiga jenis bunga, yaitu Setosa, Versicolor, dan Virginica. Dataset ini nantinya akan digunakan sebagai data untuk proses klasifikasi.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Cell ini digunakan untuk mengubah Dataset Iris ke dalam bentuk tabel menggunakan Pandas agar data lebih mudah dibaca dan diolah. Pada tabel tersebut, terdapat empat fitur, yaitu Sepal Length, Sepal Width, Petal Length, dan Petal Width, serta jenis bunga yang disimpan dalam kolom Species. Selanjutnya, data dipisahkan menjadi X yang berisi karakteristik bunga sebagai fitur dan y yang berisi jenis bunga sebagai label atau target yang akan diprediksi. Dengan pemisahan ini, data sudah siap digunakan untuk proses klasifikasi.

```python
y
```

Variabel y digunakan untuk menyimpan label atau target dari Dataset Iris, yaitu jenis bunga yang terdiri dari Setosa, Versicolor, dan Virginica. Data ini nantinya menjadi acuan bagi model dalam mempelajari dan memprediksi jenis bunga berdasarkan karakteristik yang terdapat pada variabel X.

LabelEncoder akan memberikan kode angka untuk setiap jenis bunga, sehingga Setosa, Versicolor, dan Virginica berubah menjadi label numerik. Hasil akhirnya disimpan kembali ke variabel y sebagai target yang siap digunakan dalam proses klasifikasi.

# TUGAS

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Sel ini seperti menyiapkan perlengkapan sebelum masak. numpy buat urusan hitung-hitungan angka, pandas buat bikin tabel, lalu kita ambil dataset Iris yang sudah disediakan scikit-learn. Isinya data 150 bunga iris dari 3 jenis (setosa, versicolor, virginica), lengkap dengan ukuran kelopak dan mahkotanya.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

Data mentahnya di ubah jadi tabel biar enak dilihat, mirip Excel, lalu kita tambah kolom Species yang isinya nama bunga. Setelah itu tabelnya kita pisah dua. X adalah "soal" (4 ukuran bunga: panjang dan lebar kelopak, panjang dan lebar mahkota), dan y adalah "kunci jawaban" (nama spesiesnya). Hasil print(X) menunjukkan 150 baris dan 4 kolom.

```python
y
```

Cuma buat ngecek isi y. Di sini masih berupa tulisan: setosa, versicolor, virginica.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Komputer lebih nyaman kerja dengan angka daripada tulisan. Jadi LabelEncoder mengubah setosa jadi 0, versicolor jadi 1, dan virginica jadi 2. Nanti kalau hasil prediksinya sudah keluar, kita tinggal balikin lagi ke nama aslinya.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Di bagi 150 data jadi dua kelompok. 80% (120 data) buat latihan model, dan 20% (30 data) buat ujian. Kenapa dipisah? Biar adil, karena model diuji pakai soal yang belum pernah dia lihat. random_state=0 bikin pembagian acaknya selalu sama, jadi hasil kita akan sama dengan teman-teman.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Ini bagian intinya. Di sini pakai Gaussian Naïve Bayes karena ukuran bunga itu angka kontinu (ada koma-komanya). Lewat fit(), model mempelajari 120 data latih. Dia mencatat rata-rata dan sebaran ukuran bunga untuk tiap spesies, misalnya setosa biasanya mungil dan virginica biasanya besar.

```python
result = NB_model.predict(X_test)
result
```

Sekarang 30 data ujian kita kasih ke model, dan dia menebak spesiesnya. Hasilnya berupa deretan angka 0, 1, 2 yang mewakili spesies tadi.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Tebakan model dicocokkan dengan kunci jawaban asli. Hasilnya 96,67%, artinya dari 30 soal, 29 dijawab benar dan cuma 1 yang meleset. Nilai yang bagus.

```python
#prediksi inputan data baru
input_data = [[7.3, 3.8, 4.6, 2.3]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Ada bunga baru dengan ukuran 7.3 , 3.8 , 4.6 , 2.3, dan model diminta menebak jenisnya. Hasil tebakannya masih angka, jadi inverse_transform mengubahnya balik jadi nama spesies. Nanti akan muncul tulisan merah UserWarning. Ini bukan error, itu cuma pengingat karena model dilatih dengan data yang punya nama kolom, sedangkan inputan nya berupa list biasa tanpa nama kolom.

```python
input_data = pd.DataFrame(np.array([[7.3, 3.8, 4.6, 2.3]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Ukuran bunga yang dimasukkan adalah 7.3, 3.8, 4.6, dan 2.3 untuk masing-masing fitur sepal dan petal. Nilai tersebut kemudian diproses oleh NB_model untuk menentukan jenis bunga berdasarkan pola yang sudah dipelajari sebelumnya. Hasil prediksi yang berupa angka diubah kembali menjadi nama spesies menggunakan inverse_transform, kemudian ditampilkan sebagai hasil prediksi.

# Kesimpulan
Gaussian Naïve Bayes dipakai karena semua fitur di dataset Iris berupa angka kontinu. Model dilatih dengan 120 data dan diuji dengan 30 data, dengan akurasi 96,67%, jadi cukup andal untuk mengklasifikasikan bunga iris.

Untuk data baru [7.3, 3.8, 4.6, 2.3], model memprediksi bunga tersebut termasuk species virginica. Hasil ini masuk akal karena ukuran mahkotanya besar (petal width 2,3 cm), berbeda intinya dengan setosa yang bunganya kecil.

```python
# Ikuti langkah-langkah pada modul ini, kemudian analisa korelasi antara temperature_fahrenheit, pressure_mb, dan air_quality_Sulphur_dioxide.
df_noid[['temperature_fahrenheit', 'pressure_mb',
         'air_quality_Sulphur_dioxide']].corr()
```

●	Temperature (Fahrenheit) vs Pressure (mb): nilai korelasi -0.486833 menunjukkan hubungan negatif dengan kekuatan sedang. Artinya, ketika suhu udara naik, tekanan udara cenderung menurun, dan sebaliknya.

●	Temperature (Fahrenheit) vs Sulphur Dioxide: nilai korelasi -0.023152 sangat mendekati 0, artinya praktis tidak ada hubungan linear yang berarti antara suhu udara dan kadar sulfur dioksida.

●	Pressure (mb) vs Sulphur Dioxide: nilai korelasi 0.018620 juga sangat mendekati 0, sehingga tekanan udara dan kadar sulfur dioksida juga tidak menunjukkan hubungan linear yang berarti.

Maka kesimpulan dari ketiga pasangan variabel di atas, hanya pasangan temperature_fahrenheit dan pressure_mb yang menunjukkan korelasi yang cukup berarti (negatif sedang). Sementara itu, air_quality_Sulphur_dioxide tidak berkorelasi kuat dengan temperature_fahrenheit maupun pressure_mb, sehingga kadar sulfur dioksida pada dataset ini lebih dipengaruhi oleh faktor lain di luar suhu dan tekanan udara.

```python
# Analisis di dataset shopping_trends.csv
import pandas as pd
df = pd.read_csv("shopping_trends.csv")
df
```

```python
# Analisa mean, std, min, max, Q1, Q2, Q3 untuk data Previous Purchases  di dataset shopping_trends.csv
df[['Previous Purchases']].describe()
```

Rata-rata jumlah pembelian sebelumnya (Previous Purchases) pelanggan adalah 25,35 kali, dengan simpangan baku 14,45 yang menunjukkan variasi cukup besar antar pelanggan (rentang dari 1 hingga 50 kali pembelian). Nilai Q1 = 13, Q2 (median) = 25, dan Q3 = 38 menunjukkan bahwa separuh pelanggan memiliki riwayat pembelian sebelumnya antara 13 hingga 38 kali. Karena median (25) hampir sama dengan mean (25,35), distribusi data Previous Purchases relatif simetris/merata.

```python
# Analisa korelasi Age vs Purchase Amount (USD) di dataset shopping_trends.csv
df[['Age', 'Purchase Amount (USD)']].corr()
```

Nilai korelasi antara Age dan Purchase Amount (USD) adalah -0.010424, sangat mendekati 0. Artinya usia pelanggan praktis tidak memiliki hubungan linear dengan besarnya nilai pembelian dan pelanggan yang lebih tua tidak cenderung berbelanja dengan nominal lebih besar maupun lebih kecil dibandingkan pelanggan yang lebih muda.

```python
# Analisa Korelasi Purchase Amount (USD) vs Review Rating di dataset shopping_trends.csv
df[['Purchase Amount (USD)', 'Review Rating']].corr()
```

Nilai korelasi antara Purchase Amount (USD) dan Review Rating adalah 0.030776, sangat mendekati 0. Ini menunjukkan bahwa besar-kecilnya nominal pembelian tidak berkaitan dengan tinggi-rendahnya rating ulasan yang diberikan pelanggan yang membeli dalam jumlah besar tidak serta-merta memberi rating yang lebih baik (atau lebih buruk).

## Kesimpulan
- Jumlah pembelian sebelumnya (Previous Purchases) pelanggan cukup bervariasi, dari 1 sampai 50 kali, tapi rata-ratanya di angka 25 kali dan sebarannya cukup merata.

- Usia pelanggan (Age) dan rating yang dikasih (Review Rating) sama-sama tidak berpengaruh ke besarnya uang yang dibelanjakan (Purchase Amount). Kedua korelasinya mendekati 0. Jadi kemungkinan besar, nilai pembelian pelanggan lebih dipengaruhi hal lain, misalnya jenis barang atau kategori produk yang dibeli, bukan dari usia atau puas-tidaknya pelanggan.

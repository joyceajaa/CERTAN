```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y= np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d' )
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
# Soal 2 — Grafik 2 Dimensi
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2

plt.figure(figsize=(6,6))
plt.contour(X, Y, F, levels=[1], colors='red')
plt.title("Grafik 2 Dimensi")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
# Soal 3 — Grafik 3 Dimensi
x = np.linspace(-np.pi, np.pi, 80)
y = np.linspace(-np.pi, np.pi, 80)
X, Y = np.meshgrid(x, y)

r = np.sqrt(X**2 + Y**2)
Z = np.exp(-(X**2+Y**2)) * np.sin(r)

ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='coolwarm')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```
